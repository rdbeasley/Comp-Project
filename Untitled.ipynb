```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

#parameters
N = 100
L = 1.0
eta = 0.72
rho = N/L**2
sigma = 2* np.sqrt(eta * L /(np.pi * N))
dr = sigma/20

#starting lattice
def initial_lattice(N, L):
    n = int(np.ceil(np.sqrt(N)))
    a = L/n

    positions = np.zeros((N,2))
    count = 0

    for i in range(n):
        for j in range(n):
            if count < N:
                positions[count] = [(i + 0.5) * a, (j + 0.5) * a]
                count += 1

    return positions

particle_positions = initial_lattice(N, L)

```

```python
fig, ax = plt.subplots(figsize=(6, 6))

# Plot each particle as a circle
for pos in particle_positions:
    # Circle takes (x, y) center and radius (sigma / 2)
    circle = Circle((pos[0], pos[1]), radius=sigma/2.0, 
                    edgecolor='blue', facecolor='lightblue', alpha=0.7, linewidth=1.5)
    ax.add_patch(circle)

# Set box limits and aspect ratio
ax.set_xlim(0, L)
ax.set_ylim(0, L)
ax.set_aspect('equal')  # Crucial so circles don't look like ellipses

# Add labels and title
ax.set_xlabel('X Position')
ax.set_ylabel('Y Position')
ax.set_title(f'Initial Lattice Configuration (N={N}, $\\rho$={rho})')

# Show the grid box boundary
plt.axvline(0, color='black', linestyle='--')
plt.axhline(0, color='black', linestyle='--')
plt.axvline(L, color='black', linestyle='--')
plt.axhline(L, color='black', linestyle='--')

plt.grid(True, linestyle=':', alpha=0.5)
plt.show()
```

```python


def wrap_positions(positions, L):
    return positions % L

def get_distance_matrix(positions, L):
    """
    Computes the matrix of distances between all pairs of particles 
    using the Minimum Image Convention for periodic boundaries[cite: 3].
    """
    N = len(positions)
    # Create N x N matrix to store distances
    distances = np.zeros((N, N))
    
    for i in range(N):
        for j in range(i + 1, N):
            # Compute raw coordinate differences
            dx = positions[i, 0] - positions[j, 0]
            dy = positions[i, 1] - positions[j, 1]
            
            # Apply Minimum Image Convention: wrap differences into [-L/2, L/2]
            dx = dx - L * np.round(dx / L)
            dy = dy - L * np.round(dy / L)
            
            r_ij = np.sqrt(dx**2 + dy**2)
            
            distances[i, j] = r_ij
            distances[j, i] = r_ij  # Distance matrix is symmetric
            
    return distances
```

```python
def check_overlap(particle_idx, new_pos, positions, L, sigma):

    N = len(positions)
    for j in range(N):
        if j == particle_idx:
            continue  # Don't check against itself
            
        dx = new_pos[0] - positions[j, 0]
        dy = new_pos[1] - positions[j, 1]
        
        # Apply Minimum Image Convention
        dx = dx - L * np.round(dx / L)
        dy = dy - L * np.round(dy / L)
        
        r_ij = np.sqrt(dx**2 + dy**2)
        
        if r_ij < sigma:
            return True  # Overlap found!
            
    return False  # No overlap

def run_equilibrated_monte_carlo(positions, L, sigma, d, total_steps, N_MC=1000):
    N = len(positions)
    accepted_in_block = 0
    
    for step in range(total_steps):
        # 1. Pick a random particle and propose a move
        i = np.random.randint(0, N)
        old_pos = positions[i].copy()
        
        delta_x = d * (np.random.rand() - 0.5)
        delta_y = d * (np.random.rand() - 0.5)
        new_pos = wrap_positions(old_pos + np.array([delta_x, delta_y]), L)
        
        # 2. Check overlap and update if valid
        if not check_overlap(i, new_pos, positions, L, sigma):
            positions[i] = new_pos
            accepted_in_block += 1
            
        # 3. Check and adjust step size d every N_MC steps
        if (step + 1) % N_MC == 0:
            acceptance_rate = accepted_in_block / N_MC
            
            # Adjust d based on your project guide rules
            if acceptance_rate < 0.25:
                d *= 0.95  # Decrease d by 5%
            elif acceptance_rate > 0.50:
                d *= 1.05  # Increase d by 5%
                
            print(f"Step {step+1}: Acceptance rate = {acceptance_rate*100:.1f}%, adjusted d = {d:.4f}")
            
            # Reset block counter for the next 1000 steps
            accepted_in_block = 0
            
    return positions, d


d = 0.5  # Initial maximum displacement amplitude
n_steps_test = 5000

particle_positions, acceptance_rate = run_equilibrated_monte_carlo(
    particle_positions, L, sigma, d, n_steps_test
)

print(f"Monte Carlo test complete. Acceptance rate: {acceptance_rate * 100:.2f}%")
```

```python
production_steps = 20000
sample_interval = 100  # Save configuration every 100 steps
saved_configs = []

for step in range(production_steps):
    i = np.random.randint(0, N)
    old_pos = particle_positions[i].copy()
    
    delta_x = d * (np.random.rand() - 0.5)
    delta_y = d * (np.random.rand() - 0.5)
    new_pos = wrap_positions(old_pos + np.array([delta_x, delta_y]), L)
    
    if not check_overlap(i, new_pos, particle_positions, L, sigma):
        particle_positions[i] = new_pos
        
    # Sample configurations periodically
    if (step + 1) % sample_interval == 0:
        saved_configs.append(particle_positions.copy())

print(f"Saved {len(saved_configs)} configurations for analysis.")
```

```python
def compute_pair_correlation(configurations, L, sigma, dr, r_max):
    bins = np.arange(0, r_max + dr, dr)
    n_bins = len(bins) - 1
    r_values = 0.5 * (bins[:-1] + bins[1:])
    
    total_counts = np.zeros(n_bins)
    n_configs = len(configurations)
    N = len(configurations[0])
    rho = N / (L**2)
    
    for positions in configurations:
        dist_matrix = get_distance_matrix(positions, L)
        for i in range(N):
            distances = dist_matrix[i]
            distances = distances[distances > 0]
            counts, _ = np.histogram(distances, bins=bins)
            total_counts += counts
            
    avg_counts = total_counts / (n_configs * N)
    
    g_r = np.zeros(n_bins)
    for i in range(n_bins):
        r = r_values[i]
        n_ideal = rho * 2 * np.pi * r * dr  # Ideal gas reference in 2D[cite: 2]
        g_r[i] = avg_counts[i] / n_ideal
        
    return r_values, g_r


dr = 0.1 * sigma
r_max = L / 2.0
r_vals, g_vals = compute_pair_correlation(saved_configs, L, sigma, dr, r_max)


plt.figure(figsize=(7, 5))
plt.plot(r_vals, g_vals, label=f'$\\rho$ = {rho}', color='purple', lw=2)
plt.axhline(1.0, color='gray', linestyle='--', alpha=0.7, label='Ideal Gas limit')
plt.xlabel('Distance $r / \\sigma$')
plt.ylabel('$g(r)$')
plt.title('Pair-Correlation Function')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.5)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

# ==========================================
# 1. System Parameters & Initialization
# ==========================================
N = 64           # Number of particles (e.g., 8x8 lattice)
sigma = 1.0      # Particle diameter[cite: 1]
rho = 0.4        # Number density
L = np.sqrt(N / rho)  # Box length for 2D

d = 0.2          # Initial maximum displacement amplitude[cite: 2]

def initialize_square_lattice(N, L):
    m = int(np.ceil(np.sqrt(N)))
    a = L / m
    positions = np.zeros((N, 2))
    count = 0
    for i in range(m):
        for j in range(m):
            if count < N:
                positions[count] = [(i + 0.5) * a, (j + 0.5) * a]
                count += 1
    return positions

def wrap_positions(positions, L):
    return positions % L

def get_distance_matrix(positions, L):
    N = len(positions)
    distances = np.zeros((N, N))
    for i in range(N):
        for j in range(i + 1, N):
            dx = positions[i, 0] - positions[j, 0]
            dy = positions[i, 1] - positions[j, 1]
            dx = dx - L * np.round(dx / L)
            dy = dy - L * np.round(dy / L)
            r_ij = np.sqrt(dx**2 + dy**2)
            distances[i, j] = r_ij
            distances[j, i] = r_ij
    return distances

def check_overlap(particle_idx, new_pos, positions, L, sigma):
    N = len(positions)
    for j in range(N):
        if j == particle_idx:
            continue
        dx = new_pos[0] - positions[j, 0]
        dy = new_pos[1] - positions[j, 1]
        dx = dx - L * np.round(dx / L)
        dy = dy - L * np.round(dy / L)
        if np.sqrt(dx**2 + dy**2) < sigma:
            return True
    return False

# Initialize positions
particle_positions = initialize_square_lattice(N, L)

# ==========================================
# 2. Equilibration Phase (with auto-tuning)
# ==========================================
print("Starting equilibration phase...")
equilibration_steps = 10000
N_MC = 1000
accepted_in_block = 0

for step in range(equilibration_steps):
    i = np.random.randint(0, N)
    old_pos = particle_positions[i].copy()
    
    delta_x = d * (np.random.rand() - 0.5)
    delta_y = d * (np.random.rand() - 0.5)
    new_pos = wrap_positions(old_pos + np.array([delta_x, delta_y]), L)
    
    if not check_overlap(i, new_pos, particle_positions, L, sigma):
        particle_positions[i] = new_pos
        accepted_in_block += 1
        
    if (step + 1) % N_MC == 0:
        acceptance_rate = accepted_in_block / N_MC
        if acceptance_rate < 0.25:
            d *= 0.95  # Decrease step size if acceptance is too low[cite: 3]
        elif acceptance_rate > 0.50:
            d *= 1.05  # Increase step size if acceptance is too high[cite: 2, 3]
        accepted_in_block = 0

print(f"Equilibration complete. Final optimized step size d = {d:.4f}")

# ==========================================
# 3. Production Phase & Configuration Sampling
# ==========================================
print("Starting production phase and sampling configurations...")
production_steps = 20000
sample_interval = 100  # Save configuration every 100 steps
saved_configs = []

for step in range(production_steps):
    i = np.random.randint(0, N)
    old_pos = particle_positions[i].copy()
    
    delta_x = d * (np.random.rand() - 0.5)
    delta_y = d * (np.random.rand() - 0.5)
    new_pos = wrap_positions(old_pos + np.array([delta_x, delta_y]), L)
    
    if not check_overlap(i, new_pos, particle_positions, L, sigma):
        particle_positions[i] = new_pos
        
    # Sample configurations periodically
    if (step + 1) % sample_interval == 0:
        saved_configs.append(particle_positions.copy())

print(f"Saved {len(saved_configs)} configurations for analysis.")

# ==========================================
# 4. Compute and Plot Pair-Correlation g(r)
# ==========================================
def compute_pair_correlation(configurations, L, sigma, dr, r_max):
    bins = np.arange(0, r_max + dr, dr)
    n_bins = len(bins) - 1
    r_values = 0.5 * (bins[:-1] + bins[1:])
    
    total_counts = np.zeros(n_bins)
    n_configs = len(configurations)
    N = len(configurations[0])
    rho = N / (L**2)
    
    for positions in configurations:
        dist_matrix = get_distance_matrix(positions, L)
        for i in range(N):
            distances = dist_matrix[i]
            distances = distances[distances > 0]
            counts, _ = np.histogram(distances, bins=bins)
            total_counts += counts
            
    avg_counts = total_counts / (n_configs * N)
    
    g_r = np.zeros(n_bins)
    for i in range(n_bins):
        r = r_values[i]
        n_ideal = rho * 2 * np.pi * r * dr  # Ideal gas reference in 2D[cite: 2]
        g_r[i] = avg_counts[i] / n_ideal
        
    return r_values, g_r

# Compute g(r) with dr = 0.1 * sigma (as suggested in project guide)[cite: 3]
dr = 0.1 * sigma
r_max = L / 2.0
r_vals, g_vals = compute_pair_correlation(saved_configs, L, sigma, dr, r_max)

# Plotting g(r)
plt.figure(figsize=(7, 5))
plt.plot(r_vals, g_vals, label=f'$\\rho$ = {rho}', color='purple', lw=2)
plt.axhline(1.0, color='gray', linestyle='--', alpha=0.7, label='Ideal Gas limit')
plt.xlabel('Distance $r / \\sigma$')
plt.ylabel('$g(r)$')
plt.title('Pair-Correlation Function')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.5)
plt.show()
```
